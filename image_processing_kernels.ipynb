```python
from google.colab import files
import cv2
import matplotlib.pyplot as plt

uploaded = files.upload()

filename = list(uploaded.keys())[0]

image = cv2.imread(filename)
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

plt.imshow(image_rgb)
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
import numpy as np
import cv2
import matplotlib.pyplot as plt

# Sharpening kernel
sharpen_kernel = np.array([
    [0, -1, 0],
    [-1, 5, -1],
    [0, -1, 0]
])

# Apply kernel
sharpened = cv2.filter2D(image, -1, sharpen_kernel)

# Convert to RGB
sharpened_rgb = cv2.cvtColor(sharpened, cv2.COLOR_BGR2RGB)

# Display
plt.imshow(sharpened_rgb)
plt.title("Sharpened Image")
plt.axis("off")
plt.show()
```

```python
# Blur kernel
blur_kernel = np.array([
    [1, 2, 1],
    [2, 4, 2],
    [1, 2, 1]
], dtype=np.float32) / 16

# Apply kernel
blurred = cv2.filter2D(image, -1, blur_kernel)

# Convert to RGB
blurred_rgb = cv2.cvtColor(blurred, cv2.COLOR_BGR2RGB)

# Display
plt.imshow(blurred_rgb)
plt.title("Blurred Image")
plt.axis("off")
plt.show()
```

```python
# Averaging kernel
average_kernel = np.array([
    [1, 1, 1],
    [1, 1, 1],
    [1, 1, 1]
], dtype=np.float32) / 9

# Apply kernel
averaged = cv2.filter2D(image, -1, average_kernel)

# Convert to RGB
averaged_rgb = cv2.cvtColor(averaged, cv2.COLOR_BGR2RGB)

# Display
plt.imshow(averaged_rgb)
plt.title("Averaged Image")
plt.axis("off")
plt.show()
```

```python
plt.figure(figsize=(15, 5))

plt.subplot(1, 4, 1)
plt.imshow(image_rgb)
plt.title("Original")
plt.axis("off")

plt.subplot(1, 4, 2)
plt.imshow(sharpened_rgb)
plt.title("Sharpening")
plt.axis("off")

plt.subplot(1, 4, 3)
plt.imshow(blurred_rgb)
plt.title("Blur")
plt.axis("off")

plt.subplot(1, 4, 4)
plt.imshow(averaged_rgb)
plt.title("Averaging")
plt.axis("off")

plt.show()
```
